```python
import sqlite3
from fastapi import FastAPI
from datetime import datetime
import nest_asyncio
from pyngrok import ngrok
import uvicorn
import threading

nest_asyncio.apply()
```

```python
app = FastAPI()

gold_price_per_gram = 6000
```

```python
conn = sqlite3.connect("gold.db", check_same_thread=False)
cursor = conn.cursor()

cursor.execute("""
CREATE TABLE IF NOT EXISTS gold_purchases(
id INTEGER PRIMARY KEY AUTOINCREMENT,
user_name TEXT,
gold_amount REAL,
price REAL,
date TEXT
)
""")

conn.commit()

print("Database created successfully")
```

```python
import sqlite3
from datetime import datetime

conn = sqlite3.connect("gold.db")
cursor = conn.cursor()

gold_price_per_gram = 6000

users = [
    ("Sushant", 5000),
    ("Rahul", 8000),
    ("Priya", 10000)
]

for user, amount in users:
    grams = round(amount / gold_price_per_gram, 3)

    cursor.execute(
        "INSERT INTO gold_purchases(user_name,gold_amount,price,date) VALUES(?,?,?,?)",
        (user, grams, amount, str(datetime.now()))
    )

conn.commit()
```

```python
cursor.execute("SELECT * FROM gold_purchases")

rows = cursor.fetchall()

for row in rows:
    print(row)
```

```python
@app.post("/chat")
def chat_api(data: dict):

    message = data["message"].lower()

    gold_keywords = ["gold","gold investment","buy gold","invest gold"]

    if any(keyword in message for keyword in gold_keywords):

        return {
            "answer": "Gold is considered a safe investment and protects against inflation.",
            "nudge": "You can invest in digital gold instantly using our platform. Would you like to buy digital gold?"
        }

    else:

        return {
            "answer": "I can help with investments. Ask me about gold investment."
        }
```

```python
@app.post("/buy-gold")
def buy_gold_api(data: dict):

    try:

        user_name = data.get("user_name")
        amount = data.get("amount_in_inr")

        if not user_name or not amount:
            return {"error": "user_name and amount_in_inr are required"}

        gold_grams = amount / gold_price_per_gram

        cursor.execute("""
        INSERT INTO gold_purchases(user_name,gold_amount,price,date)
        VALUES(?,?,?,?)
        """, (user_name, gold_grams, amount, str(datetime.now())))

        conn.commit()

        return {
            "message": "Digital gold purchased successfully",
            "gold_grams": round(gold_grams,3),
            "amount_paid": amount
        }

    except Exception as e:
        return {"error": str(e)}
```

```python
def run_server():
    uvicorn.run(app, host="0.0.0.0", port=8000)

thread = threading.Thread(target=run_server)
thread.start()
```
